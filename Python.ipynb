```python
# Homework :
# Write a program to find prime numbers between 1 to 30
print("Prime numbers between 1 and 30 are:")

# Loop through each number from 2 to 30 (1 is not a prime number)
for num in range(2, 31):
    is_prime = True
    
    # Check for factors from 2 up to the square root of the number
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break  # Found a factor, so it's not prime
            
    # If no factors were found, the number is prime
    if is_prime:
        print(num, end=" ")
```
