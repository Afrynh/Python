```python
# BREAK Concept
# Write A Program to print 1 to 100 but divisible by 7 and 9
i = 1
while i <= 100:
    if i%7==0 and i%9==0:
        break
    else:
        print (i)
    i = i+1
```

```python
# Write A Program to print numbers divisible by both 7 and 9
# loop from 1 to 100
## BREAK Concept
for i in range(1,101):
    if i % 7 == 0 and i % 9 == 0:
        print ('number =', i)
        break
```

```python
# Write A Program to print numbers between 1 to 100 which are not divisible by 7
for i in range (1, 101):
    if i % 7 == 0:
        continue
    print (i)
```

```python
# Write A Program to print numbers between 1 to 100 which are divisible by 7
## CONTINUE Concept
for i in range (1, 101):
    if i % 7 == 0:
        print (i)
        continue
```

```python
# PASS Concept
i = 1
while i <= 5:
    pass
    print (i)
    i = i+1
```

```python
# Write A Program to print a name without 'A'
name = 'sukanya'
i = 0
while i <= 6:
    if name [i] == 'a':
        i = i+1
        continue
    print (name[i])
    i = i+1
```
