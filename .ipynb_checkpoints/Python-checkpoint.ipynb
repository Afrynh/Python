```python
# Homework :
# Write a program to find prime numbers between 1 to 30
print("Prime numbers between 1 and 30 are:")

# Loop through each number from 2 to 30 (1 is not a prime number)
for num in range(2, 31):
    is_prime = True
    
    # Check for factors from 2 up to the square root of the number
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break  # Found a factor, so it's not prime
            
    # If no factors were found, the number is prime
    if is_prime:
        print(num, end=" ")
```

```python
class Jar:                         #(fatema's)
    def __init__(self):
        self.candies = 100


class Withdraw:
    def execute(self, jar, amount):
        if amount <= jar.candies:
            jar.candies -= amount
            print("Withdraw:", amount)
        else:
            print("Not enough candies!")

        print("Remaining candies:", jar.candies)


class Deposit:
    def execute(self, jar, amount):
        jar.candies += amount
        print("Deposit:", amount)
        print("Total candies:", jar.candies)


jar = Jar()

user1 = Withdraw()
user2 = Deposit()
user3 = Withdraw()
user4 = Deposit()

amount = int(input("User 1: Enter candies to withdraw: "))
user1.execute(jar, amount)

amount = int(input("User 2: Enter candies to deposit: "))
user2.execute(jar, amount)

amount = int(input("User 3: Enter candies to withdraw: "))
user3.execute(jar, amount)

amount = int(input("User 4: Enter candies to deposit: "))
user4.execute(jar, amount)
```

```python
class jar():
    def __init__(self):
        pass
       
    def withdraw(self):
        global Candies_left        
        withdrawn = int(input("How many candies do you want to withdraw: "))
        if withdrawn<=Candies_left:
            print("The number of candies withdrawn=",withdrawn)
            print("Therefore, total Candies left in the jar are: ",Candies_left-withdrawn)
            Candies_left= Candies_left-withdrawn
        else:
            print("You cannot remove more than", Candies_left)
            withdrawn = int(input("Please try to withdraw again: "))
            if withdrawn<=Candies_left:
                print("The number of candies withdrawn=",withdrawn)
                print("Therefore, total Candies left in the jar are: ",Candies_left-withdrawn)
                Candies_left= Candies_left-withdrawn

    def deposit(self):
        global Candies_left
   
        deposted = int(input("How many candies do you want to deposit: "))
        if deposted +Candies_left<=100:
            print("The number of candies deposited=",deposted)
            print("Therefore, total Candies left in the jar are: ",Candies_left+deposted)
            Candies_left= Candies_left+deposted
        else:
            print("You cannot deposit more than the capacity")
            deposted = int(input("Please try to deposit again: "))
            if deposted +Candies_left<=100:
                print("The number of candies deposited=",deposted)
                print("Therefore, total Candies left in the jar are: ",Candies_left+deposted)
                Candies_left= Candies_left+deposted
jar_capacity = 100
Candies_left =jar_capacity
user1=jar()
user2=jar()
user3=jar()
user4=jar()
```

# Flatten and Ravel

```python
# Flattening
# it reduces dimension to 1 dimension.
print (arr_slice)
print ('Dimension of original array is: ', arr_slice.ndim)
print ('-------------------------------------------')
print (arr_slice.flatten())
print('Dimension of flattened array is: ', arr_slice.flatten().ndim)
```
